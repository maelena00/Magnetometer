```python
import sys
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import plotly
import plotly.express as px
import ipywidgets as widgets

print("Python    ", sys.version.split()[0])
print("numpy     ", np.__version__)
print("pandas    ", pd.__version__)
print("matplotlib", matplotlib.__version__)
print("plotly    ", plotly.__version__)
print("ipywidgets", widgets.__version__)
```

```python
# A small random dataset
rng = np.random.default_rng(42)
df = pd.DataFrame({
    "x": rng.normal(0, 1, 200),
    "y": rng.normal(0, 1, 200),
    "group": rng.choice(["A", "B"], 200),
})
df.head()
```

```python
# Static plot with matplotlib
fig, ax = plt.subplots()
for name, g in df.groupby("group"):
    ax.scatter(g["x"], g["y"], label=name, alpha=0.7)
ax.legend()
ax.set_title("matplotlib works")
plt.show()
```

```python
# Interactive plot with plotly
px.scatter(df, x="x", y="y", color="group", title="plotly works")
```

```python
# Interactive slider with ipywidgets
@widgets.interact(n=(10, 200, 10))
def plot_first_n(n=100):
    plt.hist(df["x"].head(n), bins=20)
    plt.title(f"First {n} values (ipywidgets works)")
    plt.show()
```
